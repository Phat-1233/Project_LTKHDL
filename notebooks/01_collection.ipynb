# Data Collection

## 1. Subject

- The dataset represents ten years (1999 - 2008) of clinical care at 130 US hospitals and integrated delivery network. 
- Each row concerns hospital records of patients diagnosed with diabetes, who underwent laboratory, medication and stayed up to 14 days.
- The problem is to predict the readmission of the patient after being discharged. This reflects a real-world context: Hospital readmissions not only increase the managing costs for the hospitals but also impact the morbidity and mortality of the patients, who may face complications associated with diabetes.
- Therefore, determining the most likely patients to be readmitted can help the hospitals to focus on them.

## 4. Collection method
- The dataset is the clinical records of the patients collected from 130 US hospitals, not a survey.
- **Population**: Diabete patients from 130 US hospitals and integrated network delivery.
- **Period**: 10 years (1999 - 2008)
- **Limitations and biases**: The original source contains total 101766 records with 47 features. However, in the Kaggle version, the dataset is reduced to only 25000 records and 16 features. The sampling method is not publicised.  

## 5. Motivation
- **A real and costly problem**: Hospital readmissions are expensive for hospitals, insurers and patients, and they are a common indicator of care quality. In the US, the Hospital Readmissions Reduction Program (HRRP) even penalises hospitals with high readmission rates. Diabetic patients are a high-risk group because they often have comorbidities (circulatory, respiratory diseases) and take many medications.
- **Clear stakeholders**: The answers can help doctors and hospital managers identify high-risk patients at discharge and decide where to spend follow-up resources (follow-up calls, medication review, outpatient appointments).
- **Suitable for both analysis and machine learning**: The target `readmitted` (yes/no) is fairly balanced (about 47% "yes"), which suits a classification task. The other columns describe the hospital stay (`time_in_hospital`, `n_lab_procedures`, `n_procedures`, `n_medications`), the patient's history (`n_outpatient`, `n_inpatient`, `n_emergency`), the diagnoses (`diag_1`, `diag_2`, `diag_3`) and the diabetes care (`glucose_test`, `A1Ctest`, `change`, `diabetes_med`), so we can ask questions beyond a single groupby, e.g. whether the effect of an A1C test or a medication change still holds within age groups or primary diagnoses.
- **Big enough and not too clean**: With 25,000 rows and 17 columns, the dataset meets the size requirement. It still needs real preprocessing: `medical_specialty` is "Missing" in about half of the rows, the count columns are strongly right-skewed with extreme values (e.g. `n_emergency` up to 64), and age is stored as intervals.
